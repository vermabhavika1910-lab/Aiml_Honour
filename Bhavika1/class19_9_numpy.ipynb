```python
# reshape, ravel and flatten-> multi to single
a = np.arange(1,7)
b = a.reshape((2, -1))

v = b.ravel() 
v[0] = 99
print(b[0, 0])

c= b.flatten()
c[0] = 0 
print(b[0, 0])
#c vs fortran order

c_arr = np.ones((2,3), order='C')
c_arr.strides
c_arr.flags['C_CONTIGUOUS'] # True

f_arr = np.ones((2,3), order='F')
f_arr.strides
f_arr.flags['F_CONTIGUOUS'] # True
# repeating elements : repeat vs tile

arr = np.array([1, 2])
np.repeat(arr, 3) 
np.tile(arr, 3)
```

```python
# fancy indexing: take and put
arr = np.array([10, 20, 30, 40, 50])
arr.take([0,2])
print(arr)

arr1 = np.zeros(4)
arr.put([1,2],[3,4])
print(arr1)

#ufunc Methods: .add.reduce and .add.accumulate
a = np.array([1, 2, 3, 4])
np.add.reduce(a)  # 1+2+3+4

np.add.accumulate(a)  # [1, 3, 6, 10]

#trailing 
a = np.ones((4, 3))
b = np.array([1, 2, 3])
(a+b).shape

x = np.arange(4)[:, np.newaxis]
(x + b).shape

v =np.broadcast_to(b,(4, 3))
print(v.strides)
```
