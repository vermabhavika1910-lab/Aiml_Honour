```python
df = pd.DataFrame({'name': ['Alice', 'Bob', 'Charlie'], 'age': [25, 30, 35], 'salary': [50000, 60000, None],
                   'department': ['HR', 'IT', 'Finance'], 'manager': ['John', 'Jane', None], 'skills': ['Python', 'Java', 'SQL']})
```

```python
df.fillna(value={'salary': df['salary'].mean()}, inplace=True)
```

```python
df.dropna(subset=['salary'], inplace=True)
```

```python
department_df = pd.DataFrame({'id': [1, 2, 3],
                              'department': ['HR', 'IT', 'Finance'], 
                              'location': ['New York', 'San Francisco', 'Chicago']})
```

```python
df_merged = pd.merge(df, department_df, on='department', how='outer')
df_merged
```
